```python
from pathlib import Path
import json
import joblib
import numpy as np
import pandas as pd
import platform
import sklearn
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix, classification_report

X = df.drop(columns=['private']).select_dtypes(include='number')
labels = df['private'].astype('string').str.strip().str.lower()
y = labels.map({'yes': 1, 'no': 0})
valid_rows = y.notna()
if not valid_rows.all():
    print('Dropping rows with missing/unrecognized private labels:', int((~valid_rows).sum()))
X = X.loc[valid_rows]
y = y.loc[valid_rows].astype(int)
feature_columns = X.columns.tolist()
print('Target class counts (0=No, 1=Yes):')
print(y.value_counts().sort_index())

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

model_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median', keep_empty_features=True)),
    ('scaler', StandardScaler()),
    ('classifier', LogisticRegression(
        class_weight='balanced', C=1.0, max_iter=2000, random_state=42
    ))
])

model_pipeline.fit(X_train, y_train)
predictions = model_pipeline.predict(X_test)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_validate(
    model_pipeline,
    X,
    y,
    cv=cv,
    scoring={'accuracy': 'accuracy', 'precision': 'precision', 'recall': 'recall'}
)

print('Test Accuracy:', round(accuracy_score(y_test, predictions), 4))
print('Test Precision:', round(precision_score(y_test, predictions), 4))
print('Test Recall:', round(recall_score(y_test, predictions), 4))
print('CV Accuracy: %.4f (+/- %.4f)' % (scores['test_accuracy'].mean(), scores['test_accuracy'].std() * 2))
print('CV Precision: %.4f (+/- %.4f)' % (scores['test_precision'].mean(), scores['test_precision'].std() * 2))
print('CV Recall: %.4f (+/- %.4f)' % (scores['test_recall'].mean(), scores['test_recall'].std() * 2))
print('\nConfusion Matrix:')
print(confusion_matrix(y_test, predictions))
print('\nClassification Report:')
print(classification_report(y_test, predictions, target_names=['No', 'Yes']))

# Refit on all labeled rows after evaluation, then save the complete preprocessing/model pipeline.
model_pipeline.fit(X, y)
model_dir = Path.cwd() / 'model'
model_dir.mkdir(parents=True, exist_ok=True)
model_path = model_dir / 'college_private_pipeline.joblib'
metadata_path = model_dir / 'metadata.json'
joblib.dump(model_pipeline, model_path)

metadata = {
    'model_type': 'StandardScaler + LogisticRegression',
    'target_column': 'private',
    'target_mapping': {'0': 'No', '1': 'Yes'},
    'positive_class': 1,
    'feature_columns': feature_columns,
    'default_values': {name: float(value) for name, value in X.median().items()},
    'python_version': platform.python_version(),
    'numpy_version': np.__version__,
    'scikit_learn_version': sklearn.__version__,
    'pandas_version': pd.__version__,
    'joblib_version': joblib.__version__,
}
with metadata_path.open('w', encoding='utf-8') as metadata_file:
    json.dump(metadata, metadata_file, indent=2)

loaded_pipeline = joblib.load(model_path)

def predict_private(colleges: pd.DataFrame) -> pd.DataFrame:
    if not isinstance(colleges, pd.DataFrame):
        raise TypeError('Input must be a pandas DataFrame.')
    missing_features = [name for name in feature_columns if name not in colleges.columns]
    if missing_features:
        raise ValueError(f'Missing required feature columns: {missing_features}')
    features = colleges.loc[:, feature_columns]
    predicted_labels = loaded_pipeline.predict(features)
    yes_index = list(loaded_pipeline.classes_).index(1)
    yes_probability = loaded_pipeline.predict_proba(features)[:, yes_index]
    return pd.DataFrame({
        'private_prediction': pd.Series(predicted_labels, index=colleges.index).map({0: 'No', 1: 'Yes'}),
        'yes_probability': yes_probability,
    }, index=colleges.index)

print('\nSaved model pipeline:', model_path.resolve())
print('Saved model metadata:', metadata_path.resolve())
print('\nReloaded-model smoke test:')
print(predict_private(X_test.head(3)))
```
